```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sb
students = pd.read_csv("student-mat.csv", delimiter=';')
print("First 5 Records:")
print(students.head())
print("\nMissing Values:")
print(students.isna().sum())
students.drop_duplicates(inplace=True)
rows, cols = students.shape
print(f"\nRows: {rows}, Columns: {cols}")
print("\nColumn Data Types:")
print(students.info())
avg_score = students["G3"].mean()
print("\nAverage Final Grade:", round(avg_score, 2))
excellent_students = len(students[students["G3"] > 15])
print("Students Scoring Above 15:", excellent_students)
study_grade_relation = students[["studytime", "G3"]].corr().iloc[0, 1]
print("Correlation (Study Time vs Grade):", round(study_grade_relation, 3))
gender_result = students.groupby("sex").agg({"G3": "mean"})
print("\nAverage Grade by Gender:")
print(gender_result)
plt.figure(figsize=(7, 4))
plt.hist(students["G3"], bins=10)
plt.title("Grade Distribution")
plt.xlabel("G3 Score")
plt.ylabel("Student Count")
plt.show()
plt.figure(figsize=(7, 4))
plt.scatter(students["studytime"], students["G3"])
plt.title("Study Hours and Final Marks")
plt.xlabel("Study Time")
plt.ylabel("Final Grade")
plt.show()
plt.figure(figsize=(6, 4))
sb.boxplot(data=students, x="sex", y="G3")
plt.title("Gender-wise Grade Comparison")
plt.xlabel("Gender")
plt.ylabel("Final Grade")
plt.show()
print("\nAnalysis Completed Successfully.")
```
